```python
import serial
import time
import csv

# Update this to your Arduino's serial port
serial_port = '/dev/ttyUSB0'  # Change this as needed (COM3 for Windows)
baud_rate = 9600
output_file = 'data.csv'

# Create a seria
#  connection
ser = serial.Serial(serial_port, baud_rate)
time.sleep(2)  # Wait for the serial connection to establish

with open(output_file, mode='w', newline='') as csv_file:
    csv_writer = csv.writer(csv_file)

    # Read the header line (A1,A2)
    header = ser.readline().decode().strip()
    csv_writer.writerow(header.split(','))  # Write the header to the CSV file

    # Read data for 60 seconds
    start_time = time.time()
    while time.time() - start_time < 60:
        line = ser.readline().decode().strip()  # Read a line from the serial
        if line:  # If line is not empty
            csv_writer.writerow(line.split(','))  # Write the readings to the CSV file

ser.close()  # Close the serial connection
print(f"Data saved to {output_file}")
```
